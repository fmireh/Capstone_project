## Logistic Regression

```python
# Importing necessary libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn import preprocessing

# Loading the Iris dataset
iris = load_iris()
data = pd.DataFrame(iris.data, columns=iris.feature_names)
data['Species'] = iris.target
data.head()
```

```python
# Select feature columns (all except 'Species')
X = data.iloc[:,[0,1,2,3]].values

depend = data.iloc[:,4].map({
    0: 0,
    1: 1,
    2: 1
})


# used this to find out why my map was giving me Nan before --- print(data["Species"].unique())

# Select the target column ('Species')
y = depend.values

# to see what values y give me - print(y)

# Reshape X to have 4 features per row
X = X.reshape(-1, 4)
# Scale the feature values for easier model fitting
X = preprocessing.scale(X)

# Split the data into training and test sets
X_train, X_test, y_train, y_test = train_test_split(X,
                y, test_size=0.3, random_state=0)

print("Training data:",X_train.shape, y_train.shape)
print("Test data:",X_test.shape, y_test.shape)
```

```python
# Fit the logistic regression model to the training data
log_reg = LogisticRegression()
log_reg.fit(X_train, y_train)

# Make predictions on the test data and reshape the output
y_pred = log_reg.predict(X_test).reshape(-1,1)
```

```python
from sklearn.metrics import confusion_matrix

# Create a list of class names from the Iris dataset
classes = ['Iris-setosa', 'not Iris-setosa']

# Generate the confusion matrix using the true labels (y_test) and predicted labels (y_pred)
conf_mat = confusion_matrix(y_test, y_pred)

# Convert the confusion matrix into a DataFrame for better visual representation
cm_df = pd.DataFrame(conf_mat,
                     index=[f'Actual: {c}' for c in classes],
                     columns=[f'Predicted: {c}' for c in classes])

cm_df
```

```python
TP = cm_df.iloc[0, 0]
FN = cm_df.iloc[0, 1]
FP = cm_df.iloc[1, 0]
TN = cm_df.iloc[1, 1]

manual_recall = TP/(TP+FN)

manual_precision = TP/(TP+FP)

manual_f1 = (2 * manual_precision * manual_recall) / (manual_precision + manual_recall)

manual_accuracy = (TP + TN) / (TP + FN + FP + TN)

print('Recall calculated manually: {}\n'.format(manual_recall))
print('Precision calculated manually: {}\n'.format(manual_precision))
print('F1 score calculated manually: {}\n'.format(manual_f1))
print('Accuracy calculated manually: {}\n'.format(manual_accuracy))
```

The confusion matrix is based on the training data and sums to 45. The model has values for True Positives (TP) and True Negatives (TN), with no False Positives (FP) or False Negatives (FN). This means that, on the training data, the model has correctly classified all observations.

Since there are no False Positives, the precision is 100%. Since there are no False Negatives, the recall is also 100%. Therefore, the model has similar values for both precision and recall, with both being perfect on the training data.

However, because the confusion matrix is based on the training data, this does not necessarily mean the model will perform perfectly on unseen data.

```python
from sklearn.metrics import f1_score, precision_score, recall_score
# Precision and Recall
prec = precision_score(y_test, y_pred)

rec = recall_score(y_test, y_pred)

# Calculate the F1 score
f1 = f1_score(y_test, y_pred)

# Use the score method to calculate the accuracy of the logistic regression model.
score = log_reg.score(X_test, y_test)

print('Recall : {}'.format(rec))
print('Precision: {}'.format(prec))
print('F1 score: {}'.format(f1))
print('Accuracy: {}'.format(score))
```

The manually calculated accuracy, precision, and recall match the scikit-learn results, confirming the calculations are correct.
